##**Answer**


Derivation of the Bias in the MLE Covariance Matrix

We begin with the Maximum Likelihood Estimator for the population covariance matrix $\boldsymbol{\Sigma}$:


$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$


where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ is the empirical sample mean.

To determine if this estimator is biased, we expand the deviations around the true, hidden population mean $\boldsymbol{\mu}$:


$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this back into the sum inside the MLE formula gives:


$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the product matrix term-by-term yields:


$$= \sum_{i=1}^n (\mathbf{x}_i - \boldsymbol{\mu})(\mathbf{x}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{x}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{x}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Notice that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Using this simplification, the two middle cross-product terms contract:


$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Now, we apply the expectation operator $\mathbb{E}[\cdot]$ to both sides:


$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

By definition of the structural population moments:

* $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these back into our expectation equation gives:


$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

Because $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the estimator is systematically downward biased. This occurs because calculating variations against the *sample* mean rather than the *true* population mean artificially underrepresents the real spatial scattering of the system.

**Bessel's Correction Elimination of Bias**

To yield an unbiased estimator ($\mathbf{S}$), we scale the maximum likelihood matrix by the reciprocal of the bias factor ($\frac{n}{n-1}$):


$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Evaluating the expectation confirms its lack of bias:


$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

**Operational Error Diagnostics**

* **Type I Error ($\alpha$):** A **False Alarm**. This happens when the diagnostic system concludes that the structure is damaged or unstable (rejects $H_0$), when in reality the physical system remains completely nominal and healthy.
* **Type II Error ($\beta$):** A **Missed Detection**. This occurs when a structural shift or localized crack has compromised the asset (rejects $H_1$), but the diagnostic system fails to alert the team (fails to reject $H_0$).

**Consequence of an Ultra-Conservative Threshold ($\alpha = 0.0001$)**

When $\alpha$ is reduced dramatically to eliminate false alarms, the critical value boundary of the test statistic is pushed deep into the extreme tails of the baseline distribution.

1. **Impact on Statistical Power ($1-\beta$):** Because the threshold for detecting anomalies becomes incredibly stringent, minor or emerging structural shifts will fail to cross the new partition line. Therefore, $\beta$ (missed detections) will sharply escalate, causing the **statistical power ($1-\beta$) of the model to collapse**.
2. **Impact on Confidence Ellipsoids:** The confidence level of the multi-dimensional boundary scales as $1-\alpha = 0.9999$. This means the **geometric volume of the healthy operational ellipsoid expands significantly**. The boundary swells to encompass almost all variation, making the system highly insensitive to genuine baseline degradation.

The Lindeberg-Lévy Multivariate Central Limit Theorem establishes that as the sample snapshot size approaches infinity ($n \to \infty$), the standardized difference between the empirical mean and population mean converges in distribution to a Multivariate Gaussian:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

##**Answer**




### **Slutsky's Theorem (Rigorous Statement)**

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables (or vectors/matrices). If

$$X_n \xrightarrow{d} X \quad \text{and} \quad Y_n \xrightarrow{p} c \quad (c \text{ a constant})$$

then:

- $X_n + Y_n \xrightarrow{d} X + c$
- $Y_n X_n \xrightarrow{d} c X$
- $X_n / Y_n \xrightarrow{d} X / c$ provided $c \neq 0$

### Proof: Substituting $\mathbf{S}$ for $\boldsymbol{\Sigma}$ via Slutsky's Theorem

**Step 1 — Multivariate CLT:**

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

Pre-multiplying by $\boldsymbol{\Sigma}^{-1/2}$ normalises to the standard Gaussian space:

$$\boldsymbol{\Sigma}^{-1/2}\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I})$$

**Step 2 — Consistency of $\mathbf{S}$ (WLLN):**

Because $\mathbf{S}$ is an unbiased, consistent estimator of $\boldsymbol{\Sigma}$, the Weak Law of Large Numbers guarantees:

$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \implies \mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

**Step 3 — Apply Slutsky's Theorem:**

We now multiply the sequence converging *in distribution* ($\sqrt{n}(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})$) by the sequence converging *in probability* ($\mathbf{S}^{-1/2}$):

$$\mathbf{S}^{-1/2}\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2}\cdot\mathscr{N}(\mathbf{0},\boldsymbol{\Sigma}) = \mathscr{N}(\mathbf{0},\mathbf{I})$$

**Step 4 — Recover the operational distribution:**

Re-scaling back from the standardised space to the empirical matrix metric:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \sim \mathscr{N}(\mathbf{0}, \mathbf{S}) \implies \widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\!\left(\boldsymbol{\mu},\,\frac{1}{n}\mathbf{S}\right)$$

Since $\widehat{\boldsymbol{\mu}}_n \xrightarrow{p} \boldsymbol{\mu}$ we may substitute $\widehat{\boldsymbol{\mu}}_n$ for the unknown $\boldsymbol{\mu}$ in practice, giving the **operational parametric modelling distribution**:

$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\!\left(\widehat{\boldsymbol{\mu}}_n,\,\frac{1}{n}\mathbf{S}\right) \quad$$

This validates substituting $\mathbf{S}$ for $\boldsymbol{\Sigma}$ without corrupting the asymptotic stability of the diagnostic envelope.

##**Answer**



We partition the $n=5000$ strain sensor dataset into $g=5$ consecutive blocks of size $n_j=1000$ and compute **Wilks' Lambda** and its **Bartlett $\chi^2$ approximation** to test whether the joint mean vector has remained constant over time.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats
from scipy.stats import chi2

# -----------------------------------------------------------------------
# DO NOT MODIFY: Synthetic Strain Sensor Data Generator
# -----------------------------------------------------------------------
np.random.seed(42)
n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015   # Sensor 1 drift
base_data[4000:, 2] -= 0.010   # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# -----------------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g equal consecutive blocks and evaluates
    first-moment homogeneity via Wilks' Lambda and Bartlett's Chi-Square
    asymptotic transformation.

    Parameters
    ----------
    df       : pd.DataFrame of shape (n, m) containing sensor observations.
    g_chunks : Number of non-overlapping partitions (default = 5).

    Returns
    -------
    Dictionary containing:
        wilks_lambda  - Wilks' Lambda statistic (W / |B+W|)
        chi2_stat     - Bartlett chi-square approximation
        p_value       - p-value under chi2 with m*(g-1) degrees of freedom
        df_chi2       - Degrees of freedom
        W             - Within-group scatter matrix
        B             - Between-group scatter matrix
        group_means   - (g, m) array of chunk sample means
    """
    n, m = df.shape
    chunk_size = n // g_chunks

    # 1. Split into g equal consecutive blocks
    chunks = [df.iloc[i * chunk_size : (i + 1) * chunk_size].values
              for i in range(g_chunks)]

    # 2. Compute the grand (global) mean across all observations
    grand_mean = df.values.mean(axis=0)

    # 3. Calculate Within-Chunk Scatter Matrix (W) and Between-Chunk Scatter Matrix (B)
    W = np.zeros((m, m))
    B = np.zeros((m, m))
    group_means = []

    for chunk in chunks:
        nj = len(chunk)
        mu_j = chunk.mean(axis=0)
        group_means.append(mu_j)

        # Within-group: sum of squared deviations from the chunk mean
        centred = chunk - mu_j
        W += centred.T @ centred

        # Between-group: weighted squared deviation of chunk mean from grand mean
        delta = (mu_j - grand_mean).reshape(-1, 1)
        B += nj * (delta @ delta.T)

    # 4. Wilks' Lambda = |W| / |B + W|
    det_W  = np.linalg.det(W)
    det_BW = np.linalg.det(B + W)
    wilks_lambda = det_W / det_BW

    # 5. Bartlett's chi-square approximation
    df_chi2   = m * (g_chunks - 1)
    chi2_stat = -(n - 1 - (m + g_chunks) / 2) * np.log(wilks_lambda)
    p_value   = 1 - chi2.cdf(chi2_stat, df_chi2)

    return {
        "wilks_lambda"  : float(wilks_lambda),
        "chi2_stat"     : float(chi2_stat),
        "p_value"       : float(p_value),
        "df_chi2"       : int(df_chi2),
        "W"             : W,
        "B"             : B,
        "group_means"   : np.array(group_means)
    }

# Execute
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

print("=" * 60)
print("  MANOVA First-Moment Homogeneity Test Results")
print("=" * 60)
print(f"  Wilks' Lambda (Λ)      : {results['wilks_lambda']:.8f}")
print(f"  Bartlett χ² statistic  : {results['chi2_stat']:.6f}")
print(f"  Degrees of freedom     : {results['df_chi2']}")
print(f"  p-value                : {results['p_value']:.2e}")
print("=" * 60)
print()

alpha = 0.05
if results['p_value'] < alpha:
    verdict = ("REJECT H₀ at α = 0.05.\n"
               "  → The joint mean vector has SHIFTED over time.\n"
               "  → The first moment is NOT homogeneous.\n"
               "  → The injected drift (Δμ₁ = +0.015, Δμ₃ = −0.010) in the\n"
               "    final 1,000 snapshots is statistically detectable.")
else:
    verdict = ("FAIL TO REJECT H₀ at α = 0.05.\n"
               "  → No evidence of a mean shift; first moment is homogeneous.")

print(f"  Conclusion: {verdict}")
print()
print("  Group chunk means (each row = one chunk):")
print(pd.DataFrame(results['group_means'],
                   columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'],
                   index=[f'Chunk {i+1}' for i in range(5)]).round(5).to_string())
````

###  1: Covariance of the Projected Vector
We define the transformed random varable $\mathbf{Z}_i \in \mathbb{R}^{m \times 1}$ as the projection of the centered observations onto the orthogonal eigenvector basis:

$$\mathbf{Z}_i \triangleq \mathbf{P}^T \widetilde{\mathbf{X}}_i$$

To prove that this transformation completely de-correlates the feature space, we evaluate the theoretical covariance of $\mathbf{Z}_i$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}[(\mathbf{P}^T \widetilde{\mathbf{X}}_i)(\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T]$$

Utilizing the linear properties of expectation and the matrix transpose identity $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P} = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Substituting the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P}\mathbf{\Lambda}\mathbf{P}^T$ into the equation yields the definitive diagonalization proof:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T (\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) \mathbf{P} = (\mathbf{P}^T \mathbf{P}) \mathbf{\Lambda} (\mathbf{P}^T \mathbf{P}) = \mathbf{I} \mathbf{\Lambda} \mathbf{I} = \mathbf{\Lambda}$$

Because $\mathbf{\Lambda}$ is strictly diagonal, the cross-covariances between distinct principal components are exactly zero ($\mathbb{E}[Z_{i,j}Z_{i,k}] = 0$ for all $j \neq k$).

###2: Total Variance Conservation via the Trace

**Step-by-Step Formal Proof**

**Step 1: Start with the trace of the original covariance matrix.** Substitute the spectral decomposition of $\mathbf{\Sigma} = \mathbf{P}\mathbf{\Lambda}\mathbf{P}^T$ into the trace operator:
$$\text{tr}(\mathbf{\Sigma}) = \text{tr}(\mathbf{P}\mathbf{\Lambda}\mathbf{P}^T)$$

**Step 2: Apply the cyclic permutation property.** Using the identity $\text{tr}(\mathbf{A}\mathbf{B}\mathbf{C}) = \text{tr}(\mathbf{B}\mathbf{C}\mathbf{A})$, we cyclicly shift the orthogonal matrix $\mathbf{P}$ from the front to the end of the product:
$$\text{tr}(\mathbf{\Sigma}) = \text{tr}(\mathbf{\Lambda}\mathbf{P}^T\mathbf{P})$$

**Step 3: Simplify using the orthogonality property.** Because $\mathbf{P}$ is an orthogonal matrix consisting of orthonormal eigenvectors, its transpose multiplied by itself yields the identity matrix ($\mathbf{P}^T\mathbf{P} = \mathbf{I}$):
$$\text{tr}(\mathbf{\Sigma}) = \text{tr}(\mathbf{\Lambda}\mathbf{I})$$

Since multiplying any matrix by the identity matrix leaves it unchanged:
$$\text{tr}(\mathbf{\Sigma}) = \text{tr}(\mathbf{\Lambda})$$

**Step 4: Evaluate the trace of the diagonal matrix.** By definition, the trace is the sum of the diagonal elements. Because $\mathbf{\Lambda}$ is a strictly diagonal matrix containing the eigenvalues $\lambda_j$, we get:
$$\text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^{m} \lambda_j$$

By partitioning the basis matrix into $\widehat{\mathbf{P}}_k$ and $\widehat{\mathbf{P}}_{m-k}$ to isolate the first $k$ dominant components, we explicitly divide this total system variance into the principal (explained) and residual (unexplained) subspaces:

* **Marginal and Cumulative Explained Variance:** The proportion of total variance captured within the principal subspace spanned by $\widehat{\mathbf{P}}_k$ is governed by the largest $k$ eigenvalues. Accumulating these ratios yields the cumulative explained variance ratio $\Phi(k)$, which measures the informational content of the low-dimensional score vector $\mathbf{z}_{i,k}$:

$$\Phi(k) = \frac{\sum_{j=1}^k \widehat{\lambda}_j}{\sum_{i=1}^m \widehat{\lambda}_i}$$

* **Residual Unexplained Variance:** Conversely, the information discarded by omitting the remaining $m-k$ low-variance axes represents the residual unexplained variance ratio $\Psi(k)$. This metric explicitly gauges the structural scale of the unmodeled operational subspace spanned by $\widehat{\mathbf{P}}_{m-k}$, tracking the total energy remaining in the residual score vector $\mathbf{z}_{i,m-k}$:

$$\Psi(k) = 1.0 - \Phi(k) = \frac{\sum_{j=k+1}^m \widehat{\lambda}_j}{\sum_{i=1}^m \widehat{\lambda}_i}$$

Operationalizing these metrics using the empirical eigenvalues $\widehat{\lambda}_j$ provides a mathematically rigorous thresholding strategy for our subspace decomposition. By plotting the decay of $\widehat{\lambda}_j$ against $\Psi(k)$, you can isolate the precise point $k$ where the system's true deterministic mechanical behavior—reconstructed via $\widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$—transitions into the unmodeled stochastic white noise absorbed by $\widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$.

###3: Reconstruction, Residual & Pythagorean Decomposition

To compress the feature space from $m$ dimensions to a lower dimension $k$ (where $k < m$), we partition the empirical orthogonal basis matrix $\widehat{\mathbf{P}}$ into two orthogonal subspaces:$$\widehat{\mathbf{P}} = \begin{bmatrix} \widehat{\mathbf{P}}_k & \Big| & \widehat{\mathbf{P}}_{m-k} \end{bmatrix}$$where $\widehat{\mathbf{P}}_k \in \mathbb{R}^{m \times k}$ contains the first $k$ column eigenvectors corresponding to the largest $k$ eigenvalues ($\widehat{\lambda}_1 \ge \dots \ge \widehat{\lambda}_k$). This partitioning allows us to partition $\mathbf{z}\in \mathbb{R}^m$ as
$$\mathbf{z}_{i}=\begin{bmatrix}\mathbf{z}_{i,k} \\\mathbf{z}_{i,m-k}
\end{bmatrix}$$
where $\mathbf{z}_{i,k} \in \mathbb{R}^{k}$ and $\mathbf{z}_{i,m-k} \in \mathbb{R}^{m-k}.$
From
$$\mathbf{z}_i = \widehat{\mathbf{P}}^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$$
we thus have
$$(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)=\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}.$$

The partitioning of the empirical basis matrix allows us to isolate the deterministic structural dynamics from the unmodeled residual noise. By retaining only the first $k$ principal components, we define the reconstructed feature vector $\widehat{\mathbf{x}}_i \in \mathbb{R}^m$ as the projection of the low-dimensional scores back into the original measurement space:

$$\widehat{\mathbf{x}}_i \triangleq \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$$

The geometric displacement between the actual observation $\mathbf{x}_i$ and its lower-dimensional reconstruction $\widehat{\mathbf{x}}_i$ defines the residual error vector $\mathbf{e}_i \in \mathbb{R}^m$:

$$\mathbf{e}_i = \mathbf{x}_i - \widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

Because $\widehat{\mathbf{P}}_k$ and $\widehat{\mathbf{P}}_{m-k}$ span mutually orthogonal subspaces, the total energy of the centered observation vector can be cleanly decomposed via the Pythagorean identity:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\widehat{\mathbf{P}}_k \mathbf{z}_{i,k}\|^2 + \|\widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

To construct a robust diagnostics layer, we monitor changes in both the principal and residual subspaces. This is achieved using two complementary statistical indicators.

1. Hotelling's $T^2$ Statistic (Principal Subspace)

Hotelling’s $T^2$ statistic measures the Mahalanobis distance within the $k$-dimensional principal subspace. It detects significant shifts or anomalies in the nominal, high-variance operational modes of the structure:

$$T^2_i = \mathbf{z}_{i,k}^T \widehat{\mathbf{\Lambda}}_k^{-1} \mathbf{z}_{i,k} = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$$

where $\widehat{\mathbf{\Lambda}}_k = \text{diag}(\widehat{\lambda}_1, \dots, \widehat{\lambda}_k)$. Under the assumption of multivariate normality, a statistically significant inflation in $T^2_i$ indicates that the system has drifted away from its baseline operational equilibrium.

2. The $Q$ Statistic / Squared Prediction Error (Residual Subspace)

The $Q$ statistic (also known as the Squared Prediction Error, SPE) measures the magnitude of the residual vector. It quantifies how well the baseline $k$-dimensional model explains a new observation:

$$Q_i = \|\mathbf{e}_i\|^2 = \mathbf{e}_i^T \mathbf{e}_i = \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k} = \sum_{j=k+1}^m z_{i,j}^2$$

Anomalies flagged by the $Q$ statistic indicate a break in the underlying structural correlation structure. This implies that a new, unmodeled physical phenomenon—such as a localized structural crack or sensor degradation—is projecting energy into the residual noise floor.

##Event Matching and Justification
### Event A — Environmental Load Anomaly (Extreme Wind Loads): $T^2$

Extreme operational wind loads increase the *magnitude* of structural responses — larger strains, higher accelerations — but do not alter the underlying mechanical connectivity of the structure. The sensors continue to co-vary according to the same baseline correlation pattern; the observation simply drifts far from the nominal centre *along directions the PCA model already captured*. The new snapshot therefore remains well within the principal subspace (low residual reconstruction error, $Q$ stays calm), but its Mahalanobis distance from the baseline mean inflates sharply. $T^2$ is the appropriate alarm here.

### Event B — Internal Structural Fracture (Fatigue Crack): $Q$

A fatigue crack physically reroutes load pathways and alters the mechanical boundary conditions of the system. This breaks the sensor-to-sensor correlation structure that the PCA model was built on — the historical eigenvectors no longer describe how the sensors co-vary in the damaged state. When a new observation arrives, the $k$-dimensional baseline subspace can no longer reconstruct it accurately; the residual error vector $\mathbf{e}_i$ carries the signature of this unmodelled structural change, and $Q$ spikes sharply. $T^2$, which only measures movement within the now-incorrect subspace, may remain deceptively calm. The $Q$ statistic is the appropriate alarm here.


## **Part B: Numerical Verification**



***Scenario:** You are building a real-time diagnostics layer for a structural component monitored by $m = 4$ sensors. You have a baseline archive of $n = 3,000$ nominal snapshots. To establish operational boundaries, you must write an optimization pipeline that analyzes how Hotelling’s $T^2$ and the residual $Q$ statistic respond to different truncation boundaries $k \in \{1, 2, 3\}$.*

**Your Task:**
Complete the Python class method below inside Google Colab. The function must:

1. Standardize and center the incoming realizations.
2. Decompose the unbiased sample covariance matrix $\mathbf{S}$ using `np.linalg.eigh` and sort components in descending order ($\widehat{\lambda}_1 \ge \dots \ge \widehat{\lambda}_m$).
3. Loop through every possible subspace cutoff value $k \in \{1, 2, 3\}$, calculating the sample vector array for Hotelling’s $T^2$ ($T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$) and the $Q$ statistic ($Q_i = \sum_{j=k+1}^m z_{i,j}^2$).
4. Compute the expected value (empirical mean) of $T^2$ and $Q$ across all samples for each $k$.
5. Return these metrics to verify your theoretical derivations in Part A.



##**Answer**

````python
import numpy as np


class PCADiagnostics:

    def __init__(self, X):
        self.X = np.asarray(X, dtype=float)

    def compute_pca_subspace_metrics(self):

        n, m = self.X.shape

        # --------------------------------------------------
        # Step 1: Center and standardize
        # --------------------------------------------------
        mu_hat = np.mean(self.X, axis=0)

        sigma_hat = np.std(
            self.X,
            axis=0,
            ddof=1
        )

        X_std = (self.X - mu_hat) / sigma_hat

        # --------------------------------------------------
        # Step 2: Unbiased covariance matrix
        # --------------------------------------------------
        S = (X_std.T @ X_std) / (n - 1)

        eigenvalues, eigenvectors = np.linalg.eigh(S)

        idx = np.argsort(eigenvalues)[::-1]

        eigenvalues = eigenvalues[idx]
        eigenvectors = eigenvectors[:, idx]

        # PCA scores
        scores = X_std @ eigenvectors

        results = {}

        # --------------------------------------------------
        # Step 3–4
        # --------------------------------------------------
        for k in [1, 2, 3]:

            # Hotelling T²
            T2 = np.sum(
                (scores[:, :k] ** 2) /
                eigenvalues[:k],
                axis=1
            )

            # Q statistic
            Q = np.sum(
                scores[:, k:] ** 2,
                axis=1
            )

            empirical_mean_T2 = np.mean(T2)
            empirical_mean_Q = np.mean(Q)

            # --------------------------------------------------
            # Theoretical values
            # --------------------------------------------------

            theoretical_mean_T2 = k

            theoretical_mean_Q = np.sum(
                eigenvalues[k:]
            )

            results[k] = {
                "mean_T2_empirical": empirical_mean_T2,
                "mean_T2_theoretical": theoretical_mean_T2,
                "mean_Q_empirical": empirical_mean_Q,
                "mean_Q_theoretical": theoretical_mean_Q
            }

        return {
            "eigenvalues": eigenvalues,
            "results": results
        }
````

###**Example**

````python
np.random.seed(42)

n = 3000
m = 4

cov_true = np.array([
    [1.0, 0.6, 0.3, 0.1],
    [0.6, 1.0, 0.5, 0.2],
    [0.3, 0.5, 1.0, 0.4],
    [0.1, 0.2, 0.4, 1.0]
])

X = np.random.multivariate_normal(
    mean=np.zeros(m),
    cov=cov_true,
    size=n
)

model = PCADiagnostics(X)

output = model.compute_pca_subspace_metrics()

print("\nEigenvalues")
print("=" * 50)

for i, lam in enumerate(output["eigenvalues"]):
    print(f"λ{i+1} = {lam:.6f}")

print("\n")
print(
    f"{'k':>3}"
    f"{'Emp T²':>12}"
    f"{'Theory T²':>12}"
    f"{'Emp Q':>12}"
    f"{'Theory Q':>12}"
)

print("-" * 55)

for k, r in output["results"].items():

    print(
        f"{k:>3}"
        f"{r['mean_T2_empirical']:>12.6f}"
        f"{r['mean_T2_theoretical']:>12.6f}"
        f"{r['mean_Q_empirical']:>12.6f}"
        f"{r['mean_Q_theoretical']:>12.6f}"
    )
````

###**1: Fundamental Equation of Factor Analysis**

Subspace Orthogonality:** The global physical mechanics driving the common factors exert zero statistical influence over the local, channel-specific unique errors:

$$\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}_{m \times k}.$$
Under the weight of these constraints, the theoretical correlation structure of the system can be decomposed directly. Expanding the expectation of the product $\mathbf{Z}_i\mathbf{F}_i^T$ yields:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbf{I} + \mathbf{0} = \boldsymbol{\Lambda}$$

Thus, the factor loading matrix $\boldsymbol{\Lambda}$ explicitly maps the cross-correlation between the observable variables and the hidden structural modes. We evaluate the total population correlation matrix $\mathbf{R}$ by expanding the generative formula:

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T]$$

$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T]\boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Utilizing the orthogonality properties ($\mathbf{I}$, $\mathbf{0}$, and $\boldsymbol{\Psi}$), the expression cleanly reduces to the **Fundamental Equation of Factor Analysis**:

$$\mathbf{R} = \boldsymbol{\Lambda} \boldsymbol{\Lambda}^T + \boldsymbol{\Psi}.$$

Thus factor analysis reduces to a marix factorization problem.

Focusing on the $j$-th diagonal entry of the correlation matrix $\mathbf{R}$, the total unit variance of each normalized variable $Z_{i,j}$ splits into two distinct, interpretable physical spaces:

$$\mathbb{E}[Z_{i,j}^2] = \sum_{r=1}^k \lambda_{j,r}^2 + \varphi_j^2 = 1.0$$

We define these partitioned elements as:

* **Communality ($h_j^2$):** The portion of a sensor's variance explained collectively by the $k$ shared structural factors:

$$h_j^2 \triangleq \sum_{r=1}^k \lambda_{j,r}^2$$


* **Uniqueness ($\varphi_j^2$):** The remaining portion of variance completely isolated to that specific sensor, representing a combination of localized non-structural anomalies and electronic white noise:

$$\varphi_j^2 \triangleq 1.0 - h_j^2 = \mathbb{E}[\epsilon_{i,j}^2]$$



>  **Diagnostic Subspace Principles:** Because the factor model forces all inter-variable correlation to be processed exclusively through the loading matrix ($\mathbb{E}[Z_{i,j}Z_{i,l}] = \sum_{r=1}^k \lambda_{j,r}\lambda_{l,r}$ for $j \neq l$), tracking these parameters over a sequence of realizations yields localized diagnostic power:

In this scenario, its communality $h_j^2$ will drop precipitously, forcing its individual uniqueness metric $\varphi_j^2$ to spike toward $1.0$, allowing the diagnostic layer to instantly isolate sensor malfunction from true structural damage.

###**2: Varimax Rotation — Why It Preserves Communality**

### Conceptual Contrast: Raw PCA vs. Varimax-Rotated Loading Vectors

* **Raw Principal Component Loading Vector:** PCA constructs loading vectors purely to maximize global variance explanation sequentially. The first principal component loads heavily across almost all observed variables to capture the largest common variance chunk. This creates a "general factor" where individual elements are densely populated and difficult to isolate or interpret physically.
* **Varimax-Rotated Factor Loading Vector:** A Varimax rotation is an orthogonal transformation that maximizes the variance of the squared loadings within each column. Conceptually, it drives the loadings toward extreme values—either close to $\pm 1$ or close to $0$. This achieves **simple structure**, forcing each sensor variable to associate strongly with only one (or a few) latent factors, significantly clarifying their physical meaning.



### Invariance Proof under Orthogonal Rotations

Let $\mathbf{T}$ be a $k \times k$ orthogonal rotation matrix satisfying the definition identity:

$$\mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}$$

The rotated factor loading matrix is defined as:

$$\mathbf{\Lambda}^* = \mathbf{\Lambda}\mathbf{T}$$

#### 1. Invariance of the Global Covariance Approximation
Substituting the rotated loading matrix $\mathbf{\Lambda}^*$ into the global covariance approximation equation yields:

$$\mathbf{R}^* = \mathbf{\Lambda}^*(\mathbf{\Lambda}^*)^T + \mathbf{\Psi}$$

$$\mathbf{R}^* = (\mathbf{\Lambda}\mathbf{T})(\mathbf{\Lambda}\mathbf{T})^T + \mathbf{\Psi}$$

Applying the matrix transpose reversal property $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:

$$\mathbf{R}^* = \mathbf{\Lambda}(\mathbf{T}\mathbf{T}^T)\mathbf{\Lambda}^T + \mathbf{\Psi}$$

Because $\mathbf{T}\mathbf{T}^T = \mathbf{I}$:

$$\mathbf{R}^* = \mathbf{\Lambda}\mathbf{I}\mathbf{\Lambda}^T + \mathbf{\Psi} = \mathbf{\Lambda}\mathbf{\Lambda}^T + \mathbf{\Psi} = \mathbf{R}$$

**Conclusion:** The global covariance structure is completely preserved because the orthogonal transformation collapses back to the identity matrix in the middle of the quadratic product.

#### 2. Invariance of Sensor Communality ($h_j^2$)

The communality $h_j^2$ represents the proportion of variance in the $j$-th sensor explained by all common factors combined. Mathematically, it is the sum of the squared row elements of the loading matrix:

$$h_j^2 = \sum_{m=1}^{k} \lambda_{jm}^2 = [\mathbf{\Lambda}\mathbf{\Lambda}^T]_{jj}$$

When we calculate the updated communality using the rotated matrix row elements:

$$h_j^{*2} = [\mathbf{\Lambda}^*(\mathbf{\Lambda}^*)^T]_{jj} = [\mathbf{\Lambda}\mathbf{T}\mathbf{T}^T\mathbf{\Lambda}^T]_{jj} = [\mathbf{\Lambda}\mathbf{\Lambda}^T]_{jj} = h_j^2$$

**Conclusion:** Because an orthogonal matrix transformation acts as a rigid coordinate rotation in the factor subspace

###**3: Thomson's Regression Method for Latent Factor Scores**

#### 1. Joint Subspace Covariance Construction

Because the common factors $\mathbf{F}_i$ are unobservable latent parameters, they cannot be directly extracted via geometric projections like PCA. Instead, we must treat them as hidden random vectors and calculate their realization using **Thomson’s Regression Method**, which relies on the conditional expectation $\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i]$.

We construct the joint partitioned vector $\mathbf{Y}_i$:

$$\mathbf{Y}_i = [\mathbf{Z}_i^T, \mathbf{F}_i^T]^T = \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix}$$

Assuming both the observed variables $\mathbf{Z}_i$ and latent factors $\mathbf{F}_i$ are centered with a mean of zero ($\mathbb{E}[\mathbf{Z}_i] = \mathbf{0}$ and $\mathbb{E}[\mathbf{F}_i] = \mathbf{0}$), the joint population covariance matrix $\mathbf{\Sigma}_{YY}$ expands as follows:

$$\mathbf{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \mathbb{E}\left[ \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \begin{bmatrix} \mathbf{Z}_i^T & \mathbf{F}_i^T \end{bmatrix} \right] = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \end{bmatrix}$$

Using the core definition blocks of Factor Analysis:
* $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{R}$ (The visible correlation matrix)
* $\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbf{\Lambda}$ (The factor loading matrix)
* $\mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] = (\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T])^T = \mathbf{\Lambda}^T$
* $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$ (The orthogonal common factor identity constraint)

Substituting these specific blocks back into the matrix yields the exact desired layout:

$$\mathbf{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \mathbf{\Lambda} \\ \mathbf{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$



#### 2. Derivation of Thomson's Score Equation

Under the standard assumption of a Multivariate Normal distribution, the conditional mean of a partitioned system is governed by the projection identity:

$$\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \mathbf{\mu}_2 + \mathbf{\Sigma}_{21}\mathbf{\Sigma}_{11}^{-1}(\mathbf{x}_1 - \mathbf{\mu}_1)$$

Mapping this theorem directly to our factor analysis matrix partitions from $\mathbf{\Sigma}_{YY}$:
* Let $\mathbf{X}_1 = \mathbf{Z}_i \implies \mathbf{x}_1 = \mathbf{z}_i$ represent the observed variables (with mean $\mathbf{\mu}_1 = \mathbf{0}$)
* Let $\mathbf{X}_2 = \mathbf{F}_i \implies \mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \mathbf{f}_i$ represent the latent factors (with mean $\mathbf{\mu}_2 = \mathbf{0}$)

From our structured joint covariance blocks:
* The variance block of the condition is $\mathbf{\Sigma}_{11} = \mathbf{R}$
* The cross-covariance block mapping to the target slot is $\mathbf{\Sigma}_{21} = \mathbf{\Lambda}^T$

Substituting these assignments directly into the projection formula:

$$\mathbf{f}_i = \mathbf{0} + \mathbf{\Lambda}^T \mathbf{R}^{-1} (\mathbf{z}_i - \mathbf{0})$$

$$\mathbf{f}_i = \mathbf{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$

````python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================================
# COMBINED SUBSPACE ENGINES (PCA & FA PIPELINES)
# =====================================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                        titleside="top"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }



````

###  Question 1: The Total Variance Illusion

* **Part 1 (Physical Nature of Sensor 4):**

 A Uniqueness value ($\varphi^2$) approaching $100\%$ indicates that `Sensor_4` shares zero covariance with the rest of the array. * **Part 2 (The PCA Loading Contamination):**

Looking at the **Feature Loading Matrix** in the PCA dashboard, we see a striking anomaly: `PC 1` maps almost exclusively to a single variable—`Sensor_4` (indicated by the deep red block with a loading value near $1.0$). Because `Sensor_4` has a massive, unmitigated noise variance scale ($\sigma^2 \approx 2.0$), the PCA engine maximizes global variance by dedicating its entire primary axis (`PC 1`) to tracking this single noisy sensor. It completely isolates `Sensor_1`, `Sensor_2`, and `Sensor_3` to a loading of $0$ on `PC 1`.
* **Part 3 (Engineering Risk):**

This presents a significant operational hazard. If you use standard PCA for monitoring, a single blown or short-circuited instrument will hijack your most critical monitoring axis (`PC 1`), inflating its eigenvalue to $\lambda_1 \approx 2.0$. This masks the true common underlying structural dynamics of the asset and renders your primary control chart useless for catching actual systemic failures.

###  Question 2: Decoupling Structural Loading via Rotation

* **Part 1 (Mathematical Jumble vs. Simple Structure):**

Because PCA is bound to strict, orthogonal variance maximization, look at what happens to the remaining signals on `PC 2` and `PC 4`. `PC 2` and `PC 4` exhibit high loading values across *both* `Sensor_1` and `Sensor_2` simultaneously, while also bleeding into `Sensor_3`. This cross-loading represents a mathematically forced mixture of signals. Factor Analysis applies a **Varimax rotation** to eliminate this. By maximizing loading variance within columns, it completely isolates the signals: `Sensor_1` and `Sensor_2` group cleanly into `Factor 1`, while `Sensor_3` shifts entirely into `Factor 2`.
* **Part 2 (Operational Interpretability):**

The PCA loading matrix forces a complex interpretation. If a diagnostic alarm trips on `PC 2`, an engineer has to figure out a messy combination of Sensors 1, 2, and 3. With the rotated FA framework, the layout is intuitive. `Factor 1` represents the isolated physical process causing redundancy in `Sensor_1` and `Sensor_2`, while `Factor 2` maps directly to the independent process monitored by `Sensor_3`. Troubleshooting root causes becomes instantaneous.

### Question 4: Operational Trade-offs in System Health Monitoring

* **The PCA Strategy:** Tracking the main principal subspace via $T^2$ alongside the residual space via the $Q$ statistic.
* *Vulnerability:* This approach is highly vulnerable to sensor degradation. As shown in the PCA loading matrix, if a single channel (`Sensor_4`) fails high, it structurally shifts the orientation of your principal axes, dragging the calculations for the remaining healthy channels along with it.


* **The FA Strategy:** Directly monitoring the rotated Latent Factor Scores.
* *Robustness:* This approach is highly robust. Because Factor Analysis explicitly separates localized noise via the diagonal uniqueness vector ($\Psi$), any single-channel sensor failure is safely absorbed by that channel's uniqueness profile ($\psi_4^2 \rightarrow 1$). The underlying latent factors (`Factor 1` and `Factor 2`) remain clean, stable, and uncorrupted, allowing the rest of your health monitoring network to function normally.
